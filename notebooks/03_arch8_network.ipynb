```python
import matplotlib.pyplot as plt
import seaborn as sns
import json
from pathlib import Path
from ecosentry.arch8_network import run_campaign, parse_campaign_to_dict
from ecosentry.config import load_canonical_config

config = load_canonical_config()
corbett = run_campaign(config, "corbett", nodes=50)
sesha = run_campaign(config, "seshachalam", nodes=100)
sundar = run_campaign(config, "sundarbans", nodes=30)

res_dict = {
    "corbett": parse_campaign_to_dict(corbett),
    "seshachalam": parse_campaign_to_dict(sesha),
    "sundarbans": parse_campaign_to_dict(sundar)
}
Path("results").mkdir(exist_ok=True)
with open("results/arch8_report.json", "w") as f:
    json.dump(res_dict, f, indent=2)
print("Saved Arch8 Network Results")

deliveries = [corbett.overall_delivery_rate_percent, sesha.overall_delivery_rate_percent, sundar.overall_delivery_rate_percent]
labels = ["Corbett", "Seshachalam", "Sundarbans"]
plt.figure(figsize=(8,5))
sns.barplot(x=labels, y=deliveries, palette="magma")
plt.title("Delivery Rates by Campaign")
plt.ylabel("Delivery Rate (%)")
plt.ylim(0, 100)
plt.show()
```
