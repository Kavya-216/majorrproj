```python
import matplotlib.pyplot as plt
import seaborn as sns
import json, os
from pathlib import Path
from ecosentry.arch7_energy import simulate_24h, simulate_30d, run_sensitivity, EnvironmentCondition, scenario_definitions, default_energy_model
from ecosentry.config import load_canonical_config

config = load_canonical_config()
scenarios = scenario_definitions()
model = default_energy_model(config, sf=9)

day_res = simulate_24h(scenarios['corbett'], model, sf=9)
mission_res = simulate_30d(scenarios['corbett'], model, sf=9)
sens_res = run_sensitivity(scenarios['corbett'], config, baseline_sf=9)

res_dict = {
    "day": day_res,
    "mission": mission_res,
    "sensitivity": sens_res
}
Path("results").mkdir(exist_ok=True)
with open("results/arch7_report.json", "w") as f:
    json.dump(res_dict, f, indent=2)
print("Saved Arch7 Energy Results")

days = list(range(1, 31))
plt.figure(figsize=(10,5))
plt.plot(days, [x['battery_wh'] for x in mission_res["battery_trajectory"]], marker="o")
plt.title("Battery Decay over 30 Days")
plt.xlabel("Days")
plt.ylabel("Battery Remaining (Wh)")
plt.grid()
plt.show()
```
